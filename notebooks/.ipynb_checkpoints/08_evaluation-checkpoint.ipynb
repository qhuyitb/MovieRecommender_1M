```python
# Cell 3: Load data và models (sửa import và dùng proj_root)
# Tự thiết lập proj_root nếu cell môi trường chưa được chạy (để tránh NameError)
from pathlib import Path
import sys
import pandas as pd

try:
    _ = proj_root
except NameError:
    cwd = Path.cwd().resolve()
    if (cwd / 'src').exists():
        proj_root = cwd
    elif (cwd.parent / 'src').exists():
        proj_root = cwd.parent
    else:
        proj_root = cwd

# Ensure src in sys.path
src_path = proj_root / 'src'
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))

print('proj_root:', proj_root)
print('src_path exists:', src_path.exists())

# Load datasets using proj_root to avoid relative path issues
ratings = pd.read_csv(str(proj_root / 'data' / 'cleaned' / 'ratings_cleaned.csv'))
movies = pd.read_csv(str(proj_root / 'data' / 'cleaned' / 'movies_cleaned.csv'))

from content_based_recommender import ContentBasedRecommender
from svd_recommender import load_recommender as load_svd
from neural_cf_recommender import NCFRecommender
from smooth_hybrid_recommender import SmoothHybridRecommender

# Instantiate (bọc bằng try/except vì các loader có thể nặng hoặc thiếu file)
content_rec = ContentBasedRecommender()
svd_rec = load_svd()
try:
    ncf_rec = NCFRecommender.load()
except Exception as e:
    print('NCF load failed:', e)
    ncf_rec = None
try:
    hybrid_rec = SmoothHybridRecommender.load()
except Exception as e:
    print('Hybrid load failed:', e)
    hybrid_rec = None
```

```python
# Ensure helper `create_train_test_split` is available (guarded — won't overwrite)
import pandas as pd
from sklearn.model_selection import train_test_split

if 'create_train_test_split' not in globals():
    def create_train_test_split(ratings_df: pd.DataFrame,
                            test_ratio: float = 0.2,
                            min_ratings_per_user: int = 10,
                            random_state: int = 42) -> tuple:
        # Filter users có đủ ratings
        user_counts = ratings_df['userId'].value_counts()
        valid_users = user_counts[user_counts >= min_ratings_per_user].index
        filtered_df = ratings_df[ratings_df['userId'].isin(valid_users)].copy()
        print(f"Filtered users: {len(valid_users):,} (có >= {min_ratings_per_user} ratings)")
        print(f"Filtered ratings: {len(filtered_df):,}")
        train_df, test_df = train_test_split(
            filtered_df,
            test_size=test_ratio,
            random_state=random_state,
            stratify=filtered_df['userId']
        )
        print(f"Train: {len(train_df):,} ratings")
        print(f"Test: {len(test_df):,} ratings")
        return train_df, test_df

print('Helper create_train_test_split is available')
```

```python
# Cell 3: Train/test split
train_df, test_df = create_train_test_split(ratings, test_ratio=0.2)
test_users = test_df['userId'].unique()[:100]
```

```python
# Cell 4: Khởi tạo evaluator
# Import RecommenderEvaluator from src module (src is added to sys.path by earlier cell)
try:
    from recommender_evaluator import RecommenderEvaluator
    evaluator = RecommenderEvaluator(train_df, movies)
    print('Evaluator instantiated')
except Exception as e:
    print('Failed to instantiate RecommenderEvaluator:', e)
    evaluator = None
```

```python
# Cell 5: Đánh giá Content-Based
print("=" * 80)
print("ĐÁNH GIÁ MODEL 1: CONTENT-BASED RECOMMENDER")
print("=" * 80)

results_content = evaluator.evaluate_all(
    recommender=content_rec,
    recommender_type='content',
    test_users=test_users,
    test_ratings=test_df,
    k_values=[5, 10, 20],
    rating_threshold=4.0
)

evaluator.print_results_summary(results_content)
```

```python
# Cell 6: Đánh giá SVD
print("=" * 80)
print("ĐÁNH GIÁ MODEL 2: SVD RECOMMENDER")
print("=" * 80)

results_svd = evaluator.evaluate_all(
    recommender=svd_rec,
    recommender_type='svd',
    test_users=test_users,
    test_ratings=test_df,
    k_values=[5, 10, 20],
    rating_threshold=4.0
)

evaluator.print_results_summary(results_svd)
```

```python
# Cell 7: Đánh giá NCF
print("=" * 80)
print("ĐÁNH GIÁ MODEL 3: NCF RECOMMENDER")
print("=" * 80)

results_ncf = evaluator.evaluate_all(
    recommender=ncf_rec,
    recommender_type='ncf',
    test_users=test_users,
    test_ratings=test_df,
    k_values=[5, 10, 20],
    rating_threshold=4.0
)

evaluator.print_results_summary(results_ncf)
```

```python
# Cell 8: Đánh giá Hybrid
print("=" * 80)
print("ĐÁNH GIÁ MODEL 4: HYBRID RECOMMENDER")
print("=" * 80)

results_hybrid = evaluator.evaluate_all(
    recommender=hybrid_rec,
    recommender_type='hybrid',
    test_users=test_users,
    test_ratings=test_df,
    k_values=[5, 10, 20],
    rating_threshold=4.0
)

evaluator.print_results_summary(results_hybrid)
```

```python
# Cell 9: So sánh tất cả models
print("\n" + "=" * 80)
print("SO SÁNH TOÀN BỘ 4 RECOMMENDER SYSTEMS")
print("=" * 80 + "\n")

recommenders = {
    'Content-Based': content_rec,
    'SVD': svd_rec,
    'NCF': ncf_rec,
    'Hybrid': hybrid_rec
}

comparison_df = evaluator.compare_recommenders(
    recommenders=recommenders,
    test_users=test_users,
    test_ratings=test_df,
    k_values=[5, 10, 20],
    rating_threshold=4.0
)

```

```python
# Cell 10: Visualize kết quả (Optional)
import matplotlib.pyplot as plt
import seaborn as sns

# Plot comparison
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# 1. Precision@10
comparison_df['precision@10'].plot(kind='bar', ax=axes[0, 0], color='steelblue')
axes[0, 0].set_title('Precision@10', fontsize=14, fontweight='bold')
axes[0, 0].set_ylabel('Precision')
axes[0, 0].grid(alpha=0.3)

# 2. Recall@10
comparison_df['recall@10'].plot(kind='bar', ax=axes[0, 1], color='coral')
axes[0, 1].set_title('Recall@10', fontsize=14, fontweight='bold')
axes[0, 1].set_ylabel('Recall')
axes[0, 1].grid(alpha=0.3)

# 3. Coverage
comparison_df['catalog_coverage'].plot(kind='bar', ax=axes[1, 0], color='green')
axes[1, 0].set_title('Catalog Coverage', fontsize=14, fontweight='bold')
axes[1, 0].set_ylabel('Coverage')
axes[1, 0].grid(alpha=0.3)

# 4. Diversity
comparison_df['intra_list_diversity'].plot(kind='bar', ax=axes[1, 1], color='purple')
axes[1, 1].set_title('Intra-List Diversity', fontsize=14, fontweight='bold')
axes[1, 1].set_ylabel('Diversity')
axes[1, 1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig('../figures/evaluation_comparison.png', dpi=300, bbox_inches='tight')
plt.show()

print("✓ Đã lưu visualization: ../figures/evaluation_comparison.png")
```
